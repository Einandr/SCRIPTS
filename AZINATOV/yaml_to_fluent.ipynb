```python
import yaml
import os

def yaml_to_fluent(wp, material_name, output_path=None):
    
    # Формируем полный путь к YAML файлу
    yaml_path = os.path.join(wp, f"{material_name}.yaml")
    
    # Если путь для сохранения не указан, создаём в той же папке
    if output_path is None:
        output_path = os.path.join(wp, f"{material_name}.txt")
    
    # Проверяем существование YAML файла
    if not os.path.exists(yaml_path):
        print(f"Ошибка: файл {yaml_path} не найден!")
        return
    
    # Загрузка YAML файла
    with open(yaml_path, 'r', encoding='utf-8') as f:
        data = yaml.safe_load(f)
    
    # Cp
    cp_items = sorted(data['heat_capacity'].items())
    cp_str = ' '.join([f"{t} {v}" for t, v in cp_items])
    cp_cmd = f";Cp\n/define/materials/change-create/{material_name} mixture-template y piecewise-linear {len(cp_items)} {cp_str} n n n n\n"
    
    # Lambda (теплопроводность)
    lam_items = sorted(data['heat_conductivity'].items())
    lam_str = ' '.join([f"{t} {v:.6f}" for t, v in lam_items])
    lam_cmd = f";Lambda\n/define/materials/change-create/{material_name} mixture-template n y piecewise-linear {len(lam_items)} {lam_str} n n n\n"
    
    # Viscosity
    visc_items = sorted(data['viscosity'].items())
    visc_str = ' '.join([f"{t} {v:.6e}" for t, v in visc_items])
    visc_cmd = f";Viscosity\n/define/materials/change-create/{material_name} mixture-template n n y piecewise-linear {len(visc_items)} {visc_str} n n\n"
    
    # Diffusivity
    diff_items = sorted(data['mass_diffusivity'].items())
    diff_str = ' '.join([f"{t} {v:.6e}" for t, v in diff_items])
    diff_cmd = f";Diffusivity\n/define/materials/change-create/{material_name} mixture-template n n n y piecewise-linear {len(diff_items)} {diff_str} n\n"
    
    # Запись в файл
    with open(output_path, 'w', encoding='utf-8') as f:
        f.write(cp_cmd + lam_cmd + visc_cmd + diff_cmd)
    
    print(f"Готово: {output_path}")
    print(f"Материал: {material_name}")
    print(f"Источник: {yaml_path}")

# ============= ВЫЗОВ ФУНКЦИИ =============
wp = r'D:\JOB\EM-08\plate_3d\qubiq_to fluent_jou'  # УКАЖИТЕ ВАШ ПУТЬ
name = "g_qubiq_KERO"                    # УКАЖИТЕ ИМЯ МАТЕРИАЛА

yaml_to_fluent(wp, name)
```
